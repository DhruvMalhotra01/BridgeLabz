ValueError (Bad Data)

```python
try:
    # Trying to convert a word into an integer
    number = int("hello")

except Exception as e:
    print(type(e).__name__)
    print(e)
```

Example 2: FileNotFoundError (Missing File)

```python
try:
    # Trying to open a file that isn't there
    file = open("ghost_file.txt", "r")

except Exception as e:
    print(type(e).__name__)
    print(e)
```

Example 3: IndexError (Going Out of Bounds)

```python
try:
    my_list = ["apple", "banana"]
    # There is no item at index 5 (it only goes up to 1)
    print(my_list[5])

except Exception as e:
    print(type(e).__name__)
    print(e)
```

Example 4: KeyError (Missing Dictionary Key)

```python
try:
    user_data = {"name": "John", "age": 25}
    # Trying to access a key that doesn't exist
    print(user_data["email"])

except Exception as e:
    print(type(e).__name__)
    print(e)
```

Basic Syntax

```python
try:
    x = 10 / 0
    print(x)

except Exception as e:
    print(type(e).__name__)
else:
    print("Executed!!")
finally:
    print("Code ended!!")
```

Multiple Exceptions

```python
try:
    num = int(input())
    print(10 / num)

except (ValueError, ZeroDivisionError):
    print("Invalid input or division by zero.")
```

Coustom error

```python
class InsufficientBalanceError(Exception):
    pass

balance = 500

try:
    withdraw = int(input("Enter amount: "))

    if withdraw > balance:
        raise InsufficientBalanceError("Insufficient Balance!")

    balance -= withdraw
    print("Remaining Balance:", balance)

except InsufficientBalanceError as e:
    print(e)
```
