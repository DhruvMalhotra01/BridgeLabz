1. ATM Transaction Validator


```python
# Withdrawal amount must be a multiple of 100
# Account balance must never go negative
# For each transaction, print SUCCESS or FAILED


money = int(input())
n= int(input())  #number of transactions

transactions = []

for _ in  range(n):
    transactions.append(int(input()))

for transaction in transactions:
    if transaction % 100 == 0 and money - transaction >= 0:
        money -= transaction
        print("SUCCESS")
    else:
        print("FAILED")

print(money)


    
```

2. Smart Electricity Billing


```python
# Electricity bill is calculated slab-wise:
# ● First 100 units → ₹3/unit
# ● Next 100 units → ₹5/unit
# ● Remaining → ₹8/unit

# If usage > 300 units, add 10% surcharge.



units = int(input())

if units <= 100:
    bill = units * 3

elif units <= 200:
    bill = 100 * 3 + (units - 100) * 5

else:
    bill = 100 * 3 + 100 * 5 + (units - 200) * 8


    
print(int(bill * 1.1) if units > 300 else int(bill))
```

3. Password Strength Evaluator


```python
# Given a password string:
# ● Must contain at least 1 digit
# ● Must contain at least 1 uppercase
# ● Length ≥ 8

s = input()

a = (any(ch.isdigit() for ch in s))
b = (any(ch.isupper() for ch in s))
c = (any(ch.islower() for ch in s))
d = (len(s) >= 8)

print("Strong" if a and b and c and d else "Weak")
```

5. Salary Deductoin System

```python
salary = int(input())
late_days = int(input())
absent_days = int(input())
deduction = 0
if (late_days > 10):
    deduction += 0.10 * salary
elif (late_days > 5):
    deduction += 0.05 * salary

if (absent_days > 2):
    deduction += 0.05 * salary

print(int(salary - deduction))
```

6. Prime Range Analyzer

```python
# prime number between A and B
a  = int(input())
b = int(input())

def isPrime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False;
    return True;

count = 0
for i in range(a,b+1):
    if isPrime(i):
        count +=1



print(count)

```
