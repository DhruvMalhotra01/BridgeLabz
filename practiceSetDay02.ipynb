PROG 3.1: Compute Circumference and Area of Circle

```python
radius_inch = float(input("Enter the radius of the circle in inches: "))

PI = 22 / 7
radius_cm = radius_inch * 2.54

circumference = 2 * PI * radius_cm
area = PI * (radius_cm ** 2)

print(f"\nCircumference of the Circle is {circumference:.2f} cm and")
print(f"Area of the Circle is {area:.2f} sqcm")
```

PROG 8: Retail Invoicing App

```python
# PROG 8: Retail Invoicing App

# 1. Inputs
buyer_name = input("Enter Buyer Name: ")
hg_price = float(input("Enter Household Good Price: "))
hg_qty = float(input("Enter Household Good Quantity: "))
pf_price = float(input("Enter Processed Food price: "))
pf_qty = float(input("Enter Processed Food Quantity: "))

# 2. Calculations
# Household Goods: GST = 5%
hg_base_price = hg_price * hg_qty
hg_gst = hg_base_price * 0.05
hg_total = hg_base_price + hg_gst

# Processed Food: GST = 12%
pf_base_price = pf_price * pf_qty
pf_gst = pf_base_price * 0.12
pf_total = pf_base_price + pf_gst

# Grand Totals
total_amount = hg_total + pf_total
total_rounded = round(total_amount)

# 3. Formatted Output
rupee_symbol = "\u20B9"  # Indian Rupee symbol

print(f"\nBuyer Name: {buyer_name}")
print("-" * 82)
print(
    f"| {'Item Code':^17} | {'Price/Unit':^10} | {'#unit':^5} | {'Price':^10} | {'GST':^10} | {'Total w/ GST':^12} |"
)
print("-" * 82)
print(
    f"| {'Household Good':<17} | {f'Rs {int(hg_price)}':^10} | {hg_qty:^5.1f} | {f'Rs {hg_base_price:.1f}':^10} | {f'Rs {hg_gst:.1f}':^10} | {f'Rs {hg_total:.1f}':^12} |"
)
print(
    f"| {'Processed Food':<17} | {f'Rs {int(pf_price)}':^10} | {pf_qty:^5.1f} | {f'Rs {pf_base_price:.1f}':^10} | {f'Rs {pf_gst:.1f}':^10} | {f'Rs {pf_total:.1f}':^12} |"
)
print("-" * 82)
print(f"{'Total':<68} {rupee_symbol} {total_amount:.2f}")
print(f"{'Total Round':<68} {rupee_symbol} {total_rounded:.2f}")
print("-" * 82)
```
